### 4. Data Preprocessing Pipeline
We will construct a preprocessing pipeline using `scikit-learn`'s `ColumnTransformer`. This ensures that we handle missing values and feature transformations cleanly without leaking test-set information into the training set.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

# Separate target and features
X = train_df.drop(columns=['respondent_id', 'covid_vaccine'])
y = train_df['covid_vaccine']
X_test = test_df.drop(columns=['respondent_id'])

# Identify column types
numeric_cols = X.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = X.select_dtypes(include=['object']).columns.tolist()

print(f"Numerical features: {len(numeric_cols)}")
print(f"Categorical features: {len(categorical_cols)}")
```

```python
# Define preprocessing steps for numerical features (median imputation + scaling)
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Define preprocessing steps for categorical features (imputation with 'Unknown' + One-Hot Encoding)
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='Unknown')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

# Combine into a single preprocessor
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_cols),
        ('cat', categorical_transformer, categorical_cols)
    ]
)

# Fit and transform training data to verify
X_processed = preprocessor.fit_transform(X)
print(f"Processed training features shape: {X_processed.shape}")
```

### 5. Model Evaluation and Comparison
To find the most suitable model, we will split our preprocessed training data into a training set and a validation set (80/20 split) and compare three diverse classification algorithms:
1. **Logistic Regression** (Linear baseline)
2. **Random Forest Classifier** (Ensemble bagging)
3. **LightGBM Classifier** (Gradient boosting)

We will evaluate them based on the ROC-AUC score, which is ideal for binary classification tasks with slight class imbalance.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
import lightgbm as lgb
from sklearn.metrics import roc_auc_score, roc_curve, classification_report

# Split processed training data into train and validation sets
X_train, X_val, y_train, y_val = train_test_split(
    X_processed, y, test_size=0.2, random_state=42, stratify=y
)

print(f"Train size: {X_train.shape[0]}, Validation size: {X_val.shape[0]}")
```

```python
# 1. Logistic Regression
lr_model = LogisticRegression(max_iter=1000, random_state=42)
lr_model.fit(X_train, y_train)
lr_preds = lr_model.predict_proba(X_val)[:, 1]
lr_auc = roc_auc_score(y_val, lr_preds)
print(f"Logistic Regression Validation AUC: {lr_auc:.4f}")

# 2. Random Forest
rf_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train)
rf_preds = rf_model.predict_proba(X_val)[:, 1]
rf_auc = roc_auc_score(y_val, rf_preds)
print(f"Random Forest Validation AUC: {rf_auc:.4f}")

# 3. LightGBM
lgb_model = lgb.LGBMClassifier(random_state=42, verbosity=-1)
lgb_model.fit(X_train, y_train)
lgb_preds = lgb_model.predict_proba(X_val)[:, 1]
lgb_auc = roc_auc_score(y_val, lgb_preds)
print(f"LightGBM Validation AUC: {lgb_auc:.4f}")
```

```python
# Plotting ROC Curves
plt.figure(figsize=(8, 6))
for model_name, preds, auc_score in [
    ('Logistic Regression', lr_preds, lr_auc),
    ('Random Forest', rf_preds, rf_auc),
    ('LightGBM', lgb_preds, lgb_auc)
]:
    fpr, tpr, _ = roc_curve(y_val, preds)
    plt.plot(fpr, tpr, label=f"{model_name} (AUC = {auc_score:.4f})")

plt.plot([0, 1], [0, 1], 'k--', label='Random Guess')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curves Comparison on Validation Set')
plt.legend(loc="lower right")
plt.grid(True, alpha=0.3)
plt.show()
```

### 6. Detailed Model Performance Analysis
We will analyze precision, recall, F1-score, and plot the confusion matrix for the top-performing **Logistic Regression** model and compare it with the other models to understand where errors are occurring.

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Calculate class predictions (threshold = 0.5)
lr_class_preds = lr_model.predict(X_val)
rf_class_preds = rf_model.predict(X_val)
lgb_class_preds = lgb_model.predict(X_val)

print("=== Logistic Regression Classification Report ===")
print(classification_report(y_val, lr_class_preds))

print("=== Random Forest Classification Report ===")
print(classification_report(y_val, rf_class_preds))

print("=== LightGBM Classification Report ===")
print(classification_report(y_val, lgb_class_preds))
```

```python
# Plot Confusion Matrices side-by-side
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

models = [
    ('Logistic Regression', lr_model, lr_class_preds),
    ('Random Forest', rf_model, rf_class_preds),
    ('LightGBM', lgb_model, lgb_class_preds)
]

for i, (name, model, preds) in enumerate(models):
    cm = confusion_matrix(y_val, preds)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['No Vaccine', 'Vaccine'])
    disp.plot(ax=axes[i], cmap='Blues', values_format='d')
    axes[i].set_title(f"{name} CM")

plt.tight_layout()
plt.show()
```

### 5.5 Expanded Model Comparison: Adding XGBoost and SVM
To ensure we have tested all powerful model families, we will expand our search space by training an **XGBoost Classifier** and a **Support Vector Machine (SVC)**, and compare their performance against our existing models.

```python
import xgboost as xgb
from sklearn.svm import SVC

# 4. XGBoost Classifier
xgb_model = xgb.XGBClassifier(random_state=42, use_label_encoder=False, eval_metric='logloss')
xgb_model.fit(X_train, y_train)
xgb_preds = xgb_model.predict_proba(X_val)[:, 1]
xgb_auc = roc_auc_score(y_val, xgb_preds)
print(f"XGBoost Validation AUC: {xgb_auc:.4f}")

# 5. Support Vector Classifier (Probability calibration required for ROC-AUC)
svc_model = SVC(probability=True, random_state=42)
svc_model.fit(X_train, y_train)
svc_preds = svc_model.predict_proba(X_val)[:, 1]
svc_auc = roc_auc_score(y_val, svc_preds)
print(f"SVM Validation AUC: {svc_auc:.4f}")
```

```python
# Compare all 5 models visually
plt.figure(figsize=(10, 8))
for model_name, preds, auc_score in [
    ('Logistic Regression', lr_preds, lr_auc),
    ('Random Forest', rf_preds, rf_auc),
    ('LightGBM', lgb_preds, lgb_auc),
    ('XGBoost', xgb_preds, xgb_auc),
    ('SVM', svc_preds, svc_auc)
]:
    fpr, tpr, _ = roc_curve(y_val, preds)
    plt.plot(fpr, tpr, label=f"{model_name} (AUC = {auc_score:.4f})")

plt.plot([0, 1], [0, 1], 'k--', label='Random Guess')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curves Comparison: All Models')
plt.legend(loc="lower right")
plt.grid(True, alpha=0.3)
plt.show()
```

### 6.5 Hyperparameter Tuning for the Best Model
We will use `GridSearchCV` to tune the hyperparameters of our Logistic Regression model. We will explore different regularization penalties (`l1` and `l2`), different strengths (`C`), and robust solvers like `saga` to handle both penalties cleanly.

```python
from sklearn.model_selection import GridSearchCV

# Define the parameter grid to search over
param_grid = {
    'penalty': ['l1', 'l2'],
    'C': [0.001, 0.01, 0.1, 1, 10, 100],
    'solver': ['saga'] # 'saga' supports both l1 and l2 penalties and scales well
}

# Initialize Logistic Regression with a high max_iter for convergence with SAGA solver
lr_base = LogisticRegression(max_iter=5000, random_state=42)

# Use GridSearchCV on the processed training set
# Optimizing for 'roc_auc' as our evaluation metric
grid_search = GridSearchCV(
    estimator=lr_base,
    param_grid=param_grid,
    scoring='roc_auc',
    cv=5,
    n_jobs=-1,
    verbose=1
)

print("Fitting GridSearchCV on the processed training data...")
grid_search.fit(X_train, y_train)

print(f"Best Hyperparameters: {grid_search.best_params_}")
print(f"Best Cross-Validation ROC-AUC: {grid_search.best_score_:.4f}")
```

```python
# Evaluate the tuned model on our validation set
best_lr_model = grid_search.best_estimator_
tuned_lr_preds = best_lr_model.predict_proba(X_val)[:, 1]
tuned_lr_class_preds = best_lr_model.predict(X_val)

tuned_auc = roc_auc_score(y_val, tuned_lr_preds)
print(f"Baseline Logistic Regression Validation AUC: {lr_auc:.4f}")
print(f"Tuned Logistic Regression Validation AUC: {tuned_auc:.4f}")
print("\n=== Tuned Logistic Regression Classification Report ===")
print(classification_report(y_val, tuned_lr_class_preds))
```

### 7. Generating and Saving Test Predictions
We will now use our final trained Logistic Regression model to make predictions on the test set (`dataset_C_testing.csv`). The pipeline will transform the test features automatically, ensuring consistent scaling and encoding.

```python
# Preprocess the test features using the fitted pipeline
X_test_processed = preprocessor.transform(X_test)

# Generate predictions using our tuned Logistic Regression model
test_probs = best_lr_model.predict_proba(X_test_processed)[:, 1]
test_preds = best_lr_model.predict(X_test_processed)

# Construct the final submission DataFrame
submission_df = pd.DataFrame({
    'respondent_id': test_df['respondent_id'],
    'covid_vaccine_probability': test_probs,
    'covid_vaccine_prediction': test_preds
})

# Save the results to the submission CSV file
output_path = '/content/dataset_C_predictions.csv'
submission_df.to_csv(output_path, index=False)
print(f"Saved optimized predictions using the tuned model to {output_path}")
display(submission_df.head(10))
```
